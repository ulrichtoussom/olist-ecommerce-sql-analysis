### Quelques utiles pour l'exploration de données avec pandas

````python
def display_structure(df) :

    pd_series = df.dtypes
    df_structure = pd.DataFrame(pd_series, columns=['Data Type']).reset_index()
    df_structure.rename(columns={'index': 'Column Name'}, inplace=True)
    return df_structure


def missing_values(df):
    missing_values = df.isnull().sum()
    missing_values_df = pd.DataFrame(missing_values, columns=['Missing Values']).reset_index()
    missing_values_df.rename(columns={'index': 'Column Name'}, inplace=True)
    return missing_values_df

def duplicate_values(df):
    initial_count = len(df)
    duplicate_count = df.duplicated().sum()
    duplicate_percentage = (duplicate_count / initial_count) * 100
    return {
        'name': df.name if hasattr(df, 'name') else 'DataFrame',
        'Initial Count': initial_count,
        'Duplicate Count': int(duplicate_count),
        'Duplicate Percentage': f"{int(duplicate_percentage)}%"
    }
````
