```python
import pandas as pd
```

```python
# the function that check de duplicated values in the dataframe and return the duplicated rows
def check_duplicates(df):
    return df[df.duplicated(keep=False)]

# summary of row of the dataframe and the number of duplicated rows
def summary_duplicates(df):
    total_rows = df.shape[0]
    duplicated_rows = check_duplicates(df).shape[0]
    return total_rows, duplicated_rows
def check_missing_values(df):
    return df.isnull().sum()
```

```python
df_customer = pd.read_csv("../data/olist_customers_dataset.csv")
df_geolocation = pd.read_csv("../data/olist_geolocation_dataset.csv")
df_order_items = pd.read_csv("../data/olist_order_items_dataset.csv")
df_order_payments = pd.read_csv("../data/olist_order_payments_dataset.csv")
df_orders = pd.read_csv("../data/olist_orders_dataset.csv")
df_products = pd.read_csv("../data/olist_products_dataset.csv")
df_sellers = pd.read_csv("../data/olist_sellers_dataset.csv")
df_category_name_translation = pd.read_csv("../data/product_category_name_translation.csv")
```

```python
print(check_missing_values(df_customer).sum())
print(check_missing_values(df_geolocation).sum())
print(check_missing_values(df_order_items).sum())
print(check_missing_values(df_order_payments).sum())
print(check_missing_values(df_orders).sum())  
print(check_missing_values(df_products).sum())
print(check_missing_values(df_sellers).sum())
print(check_missing_values(df_category_name_translation).sum())
```

```python
df_client = pd.DataFrame({
    "Amount": [100, 200, 300, 400, 500, 600, 700, 800, 900, 1000, 1100, 1200, 1300, 1400, 1500, 1600, 1700, 1800, 1900, 100000]
})

df_client.head(5)
```

```python
Mean = df_client['Amount'].mean()
print(Mean)
```

```python
std = df_client['Amount'].std()
print(std)
```

```python
describe_info = df_client.describe()
describe_info
```

```python
describe_info = df_client.describe()
mean = describe_info.loc['mean', 'Amount']
std = describe_info.loc['std', 'Amount']

coefficient_of_variation = std / mean
print(coefficient_of_variation)
```

Nous avons une moyenne 
 mean = 5950
Et une mediane=1050.000000

la moyenne et la mediane sont tres eloignés. ceci donne deja une idee sur la nature de la repartition des données qui semble etre asymetrique positif (dirigé vers la droite), en effet il semble avoir des valeur extreme qui ont tiré la moyenne vers le haut 

De plus nous avons un coeficient de variation largement superieur a 1 ce qui informe sur grande  dispersion des donnees autour de la moyenne .
